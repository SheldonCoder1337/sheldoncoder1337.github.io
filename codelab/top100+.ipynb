> 算法的本质是“如何聪明地遍历”
> 我们可以把算法题理解为：在给定的数据结构上，通过某种遍历策略，收集或计算出所需的信息，同时尽可能减少时间和空间消耗。遍历是必然的，但优化的空间在于：
> - 能否跳过不可能产生解的部分？——如二分查找、剪枝。
> - 能否利用历史信息避免重复遍历？——如动态规划、记忆化搜索。
> - 能否改变数据结构让遍历更高效？——如排序后使用双指针。

## Part 1: Basic data structure: array & linked list

```python
from typing import List, Optional

class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random

class Solution:

    """ Hash """

    def twoSum(self, nums: List[int], target: int) -> List[int]:
        """ 1
        https://leetcode.cn/problems/two-sum/description/?envType=study-plan-v2&envId=top-100-liked
        """
        hashmap = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in hashmap:
                return [hashmap[complement], i]
            hashmap[num] = i
        return []

    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        """ 49 for for
        https://leetcode.cn/problems/group-anagrams/?envType=study-plan-v2&envId=top-100-liked
        """
        from collections import defaultdict
        group = defaultdict(list)
        for s in strs:
            count = [0]*26
            for c in s:
                count[ord(c) - ord('a')] += 1 # ord
            key = tuple(count)
            group[key].append(s)
        return list(group.values())

    def longestConsecutive(self, nums: List[int]) -> int:
        """ 128 for while
        https://leetcode.cn/problems/longest-consecutive-sequence/?envType=study-plan-v2&envId=top-100-liked
        """
        if not nums:
            return 0
        num_set = set(nums) # set
        longest_length = 0
        for num in num_set:
            if num-1 not in num_set: # the first num of the sequence
                current_num = num
                current_length = 1
                while current_num + 1 in num_set:
                    current_num += 1
                    current_length += 1
                longest_length = max(longest_length, current_length)
        return longest_length

    """ Double Pointer """

    def moveZeroes(self, nums: List[int]) -> None:
        """ 283
        https://leetcode.cn/problems/move-zeroes/?envType=study-plan-v2&envId=top-100-liked
        """
        j = 0
        for i in range(len(nums)):
            if nums[i] != 0:
                nums[j] = nums[i]
                j += 1
        for k in range(j, len(nums)):
            nums[k] = 0

    def maxArea(self, height: List[int]) -> int:
        """ 11
        https://leetcode.cn/problems/container-with-most-water/?envType=study-plan-v2&envId=top-100-liked
        """
        left, right = 0, len(height)-1
        max_area = 0
        while left < right:
            current_area = (right-left) * min(height[left], height[right])
            max_area = max(max_area, current_area)
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return max_area
    
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        """ 15 for while
        https://leetcode.cn/problems/3sum/?envType=study-plan-v2&envId=top-100-liked
        """
        nums.sort() # TODO: logn
        n = len(nums)
        res = []
        for i in range(n):
            if nums[i] > 0:
                break
            if nums[i] == nums[i-1] and i > 0:
                continue

            left, right = i+1, n-1
            while left < right:
                total = nums[i] + nums[left] + nums[right]
                if total == 0:
                    res.append([nums[i], nums[left], nums[right]])

                    while left<right and nums[left] == nums[left+1]:
                        left += 1
                    while left<right and nums[right] == nums[right-1]:
                        right -= 1

                    left += 1
                    right -= 1
                elif total<0:
                    left += 1
                else:
                    right -= 1
        return res 

    # NOTE
    def trap(self, height: List[int]) -> int:
        """ 42
        https://leetcode.cn/problems/trapping-rain-water/description/?envType=study-plan-v2&envId=top-100-liked
        
        接雨水问题要求计算柱状图在雨后能储存的总水量。每个位置能接的雨水量取决于它左右两侧最高柱子中的较矮者减去自身高度。我们可以使用双指针法从两端向中间遍历, 动态维护左右两侧的最大高度, 并累加每个位置的水量。
        核心思想：对于当前指针 left 和 right, 如果 height[left] < height[right], 则 left 位置的水量由左侧最高 left_max 决定（因为右侧更高, 能挡住水）；反之则由右侧最高 right_max 决定。
        通过比较两端高度, 逐步向中间收缩, 保证每个位置的水量被正确计算。
        """
        if not height:
            return 0
        left, right = 0, len(height) - 1
        left_max = right_max = ans = 0
        while left < right:
            if height[left] < height[right]:
                if height[left] >= left_max:
                    left_max = height[left]
                else:
                    ans += left_max - height[left]
                left += 1
            else:
                if height[right] >= right_max:
                    right_max = height[right]
                else:
                    ans += right_max - height[right]
                right -= 1
        return ans

    """ Sliding window """

    def lengthOfLongestSubstring(self, s: str) -> int:
        """ 3 for while
        https://leetcode.cn/problems/longest-substring-without-repeating-characters/description/?envType=study-plan-v2&envId=top-100-liked
        """
        occ = set() # occurrence
        n = len(s)
        right = -1
        max_length = 0

        for left in range(n):
            if left != 0:
                occ.remove(s[left-1])
            while right+1<n and s[right+1] not in occ:
                occ.add(s[right+1])
                right += 1
            max_length = max(max_length, right-left+1)
        return max_length
    
    def findAnagrams(self, s: str, p: str) -> List[int]:
        """ 438
        https://leetcode.cn/problems/find-all-anagrams-in-a-string/description/?envType=study-plan-v2&envId=top-100-liked
        """
        m, n = len(s), len(p)
        if n > m:
            return []
        
        p_count = [0] * 26
        window_count = [0] * 26
        for ch in p:
            p_count[ord(ch) - ord('a')] += 1
        
        for i in range(n):
            window_count[ord(s[i]) - ord('a')] += 1
        res = []
        if window_count == p_count:
            res.append(0)

        for i in range(n, m):
            window_count[ord(s[i]) - ord('a')] += 1
            window_count[ord(s[i-n]) - ord('a')] -= 1
            if window_count == p_count:
                res.append(i-n+1)

        return res   

    """ subarray """

    # NOTE
    def subarraySum(self, nums: List[int], k: int) -> int:
        """ 560 前缀和
        https://leetcode.cn/problems/subarray-sum-equals-k/?envType=study-plan-v2&envId=top-100-liked
        
        核心思想：
        定义前缀和 pre[i] 表示从 nums[0] 到 nums[i-1] 的和(即前 i 个元素的和）, 则子数组 [j, i] 的和为 pre[i+1] - pre[j]。
        我们需要统计满足 pre[i+1] - pre[j] = k 的 (j, i) 对数, 其中 0 <= j <= i。
        
        !!! 等价于：对于每个 i, 统计之前有多少个 pre[j] 等于 pre[i+1] - k。

        具体做法：

        初始化一个哈希表 count, 用于存储前缀和出现的次数。

        初始化 pre = 0 表示当前前缀和（从开始到当前元素的和）。

        遍历数组 nums:

        将当前元素 num 加到 pre 上。

        在哈希表中查找 pre - k, 若存在, 则累加其出现次数到答案中: 因为 pre - k 表示存在某个前缀和, 使得当前前缀和减去该前缀和等于 k, 即中间这一段子数组和为 k。

        将当前 pre 存入哈希表(s次数+1)。

        返回答案。
        
        """
        count = {0:1}
        pre = 0
        ans = 0
        for num in nums:
            pre += num
            ans += count.get(pre-k, 0)
            count[pre] = count.get(pre, 0)+1
        return ans

    # NOTE
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        """ 239 for while
        https://leetcode.cn/problems/sliding-window-maximum/?envType=study-plan-v2&envId=top-100-liked
        """
        from collections import deque
        dq = deque()
        res = []
        for i in range(len(nums)):
            if dq and dq[0] == i - k:
                dq.popleft()
            while dq and nums[dq[-1]] <= nums[i]:
                dq.pop()
            dq.append(i)
            if i >= k - 1:
                res.append(nums[dq[0]])
        return res  

    # NOTE
    def minWindow(self, s: str, t: str) -> str:
        """ 76: for while 先从右指针开始扫，然后收缩左指针
        https://leetcode.cn/problems/minimum-window-substring/description/?envType=study-plan-v2&envId=top-100-liked
        """

        if not s or not t:
            return ""
        
        need = {}
        for ch in t:
            need[ch] = need.get(ch, 0) + 1
        need_cnt = len(need)
        
        window = {}
        have = 0
        left = 0
        min_len = float('inf')
        start = 0

        # 右指针开始扫描
        for right, ch in enumerate(s):
            window[ch] = window.get(ch, 0) + 1
            if ch in need and window[ch] == need[ch]:
                have += 1
            
            # 满足条件
            while have == need_cnt:
                # 更新最小窗口
                cur_len = right - left + 1
                if cur_len < min_len:
                    min_len = cur_len
                    start = left
                
                # 移出左边界字符
                left_char = s[left]
                window[left_char] -= 1
                if left_char in need and window[left_char] < need[left_char]:
                    have -= 1
                left += 1
        
        return "" if min_len == float('inf') else s[start:start+min_len]
    
    """ array """
    
    def maxSubArray(self, nums: List[int]) -> int:
        """ 53
        https://leetcode.cn/problems/maximum-subarray/?envType=study-plan-v2&envId=top-100-liked
        """
        current_sum = max_sum = nums[0]
        for num in nums[1:]:
            current_sum = max(num, current_sum + num)
            max_sum = max(max_sum, current_sum)
        return max_sum
    
    # NOTE
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        """ 56
        https://leetcode.cn/problems/merge-intervals/?envType=study-plan-v2&envId=top-100-liked
        """
        if not intervals:
            return []
        
        intervals.sort(key=lambda x: x[0])
        merged = [intervals[0]]
        for start, end in intervals[1:]:
            last_start, last_end = merged[-1]
            if start <= last_end:
                merged[-1][1] = max(last_end, end)
            else:
                merged.append([start, end])
        
        return merged

    def rotate(self, nums: List[int], k: int) -> None:
        """ 189
        Do not return anything, modify nums in-place instead.
        https://leetcode.cn/problems/rotate-array/description/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(nums)
        k %= n
        def reverse(start, end):
            while start<end:
                nums[start], nums[end] = nums[end], nums[start]
                start += 1
                end -= 1
        reverse(0,n-1)
        reverse(0,k-1)
        reverse(k,n-1)

    # NOTE
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        """ 238 向右扫，然后向左扫
        https://leetcode.cn/problems/product-of-array-except-self/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(nums)
        answer = [1] * n
        R = 1

        for i in range(1, n):
            answer[i] = answer[i-1] * nums[i-1]
        
        for i in range(n-1, -1, -1):
            answer[i] *= R
            R *= nums[i]
        
        return answer

    # NOTE
    def firstMissingPositive(self, nums: List[int]) -> int:    
        """ 41 要找缺失的最小正整数，答案一定在 [1, n+1] 范围内，其中 n 是数组长度。
        https://leetcode.cn/problems/first-missing-positive/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(nums)
        i = 0
        while i<n:
            if 1 <= nums[i] <= n and nums[i] != nums[nums[i] - 1]:
                target_id = nums[i]-1
                nums[i], nums[target_id] = nums[target_id], nums[i]
            else:
                i += 1
        for i in range(n):
            if nums[i] != i+1:
                return i+1
        return n+1
        
    """ matrix """

    def setZeroes(self, matrix: List[List[int]]) -> None:
        """ 73
        https://leetcode.cn/problems/set-matrix-zeroes/?envType=study-plan-v2&envId=top-100-liked
        Do not return anything, modify matrix in-place instead.
        """

        row, column = len(matrix), len(matrix[0])

        first_row_has_zero = any(matrix[0][i]==0 for i in range(column))
        first_col_has_zero = any(matrix[j][0]==0 for j in range(row))

        for i in range(1, row):
            for j in range(1, column):
                if matrix[i][j] == 0:
                    matrix[i][0] = 0
                    matrix[0][j] = 0

        for i in range(1, row):
            for j in range(1, column):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0

        if first_row_has_zero:
            for j in range(column):
                matrix[0][j] = 0
        
        if first_col_has_zero:
            for i in range(row):
                matrix[i][0] = 0

    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        """ 54 top + bottom + left + right
        https://leetcode.cn/problems/spiral-matrix/?envType=study-plan-v2&envId=top-100-liked
        """
        if not matrix or not matrix[0]:
            return []

        row, column = len(matrix), len(matrix[0])
        top, bottom = 0, row-1
        left, right = 0, column-1
        res = []

        while top<=bottom and left<=right:
            for i in range(left, right+1):
                res.append(matrix[top][i])
            top += 1

            for i in range(top, bottom+1):
                res.append(matrix[i][right])
            right -= 1

            if top<=bottom:
                for i in range(right, left-1, -1):
                    res.append(matrix[bottom][i])
                bottom -= 1

            if left<=right:
                for i in range(bottom, top-1, -1):
                    res.append(matrix[i][left])
                left += 1
        
        return res

    def rotate(self, matrix: List[List[int]]) -> None:
        """ 48 等价
        Do not return anything, modify matrix in-place instead.
        https://leetcode.cn/problems/rotate-image/description/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(matrix)
        for i in range(n):
            for j in range(i + 1, n):
                matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
        for i in range(n):
            matrix[i].reverse()
        
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        """ 240 右上角
        https://leetcode.cn/problems/search-a-2d-matrix-ii/?envType=study-plan-v2&envId=top-100-liked
        """
        if not matrix or not matrix[0]:
            return False
        
        row, col = len(matrix), len(matrix[0])

        i, j = 0, col-1
        while i < row and j >= 0:
            if matrix[i][j] == target:
                return True
            elif matrix[i][j] > target:
                j -= 1
            else:
                i += 1
        return False
            
    """ Link list """

    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        """ 160 追及问题
        https://leetcode.cn/problems/intersection-of-two-linked-lists/?envType=study-plan-v2&envId=top-100-liked
        """
        if not headA or not headB:
            return None
        pA, pB = headA, headB
        while pA is not pB:
            pA = pA.next if pA else headB
            pB = pB.next if pB else headA
        return pA
    
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """ 206
        https://leetcode.cn/problems/reverse-linked-list/?envType=study-plan-v2&envId=top-100-liked
        """
        prev = None
        curr = head
        while curr:
            next_temp = curr.next
            curr.next = prev
            prev = curr
            curr = next_temp
        return prev

    # NOTE: 快慢指针找链表中点, 反转链表，判断是否一致
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        """ 234
        https://leetcode.cn/problems/palindrome-linked-list/?envType=study-plan-v2&envId=top-100-liked
        """
        # 快慢指针找到中点
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        
        # 反转后半部分
        prev = None
        while slow:
            nxt = slow.next
            slow.next = prev
            prev = slow
            slow = nxt
        
        # 比较前半部分和后半部分
        left, right = head, prev
        while right:
            if left.val != right.val:
                return False
            left = left.next
            right = right.next
        return True
    
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        """ 141
        https://leetcode.cn/problems/linked-list-cycle/?envType=study-plan-v2&envId=top-100-liked
        """
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                return True
        return False

    # NOTE: 巧解
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """ 142
        https://leetcode.cn/problems/linked-list-cycle-ii/?envType=study-plan-v2&envId=top-100-liked
        """
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                slow = head
                while slow != fast:
                    slow = slow.next
                    fast = fast.next
                return slow
        return None   

    # NOTE: dummy = ListNode() 简化头节点处理技巧
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        """ 21
        https://leetcode.cn/problems/merge-two-sorted-lists/?envType=study-plan-v2&envId=top-100-liked
        """
        dummy = ListNode()
        cur = dummy
        while list1 and list2:
            if list1.val < list2.val:
                cur.next = list1
                list1 = list1.next
            else:
                cur.next = list2
                list2 = list2.next
            cur = cur.next
        cur.next = list1 if list1 else list2
        return dummy.next

    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        """ 2
        https://leetcode.cn/problems/add-two-numbers/?envType=study-plan-v2&envId=top-100-liked
        """
        dummy = ListNode()
        cur = dummy
        carry = 0  # 进位
        while l1 or l2 or carry:
            total = carry
            if l1:
                total += l1.val
                l1 = l1.next
            if l2:
                total += l2.val
                l2 = l2.next
            carry = total // 10  # 新的进位
            cur.next = ListNode(total % 10)  # 当前位
            cur = cur.next
        return dummy.next

    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        """ 19
        https://leetcode.cn/problems/remove-nth-node-from-end-of-list/?envType=study-plan-v2&envId=top-100-liked
        """
        dummy = ListNode(0, head)
        slow, fast = dummy, head
        for _ in range(n):
            fast = fast.next
        while fast:
            fast = fast.next
            slow = slow.next
        slow.next = slow.next.next
        return dummy.next

    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """ 24
        https://leetcode.cn/problems/swap-nodes-in-pairs/description/?envType=study-plan-v2&envId=top-100-liked
        """
        dummy = ListNode(0, head)
        prev = dummy
        while prev.next and prev.next.next:
            first = prev.next
            second = prev.next.next
            first.next = second.next
            second.next = first
            prev.next = second
            prev = first
        return dummy.next

    # NOTE: 多加练习
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        """ 25
        https://leetcode.cn/problems/reverse-nodes-in-k-group/?envType=study-plan-v2&envId=top-100-liked
        """

        def reverse(start: ListNode, end: ListNode, next_group: ListNode) -> ListNode:
            prev = None
            curr = start
            while curr != next_group:
                nxt = curr.next
                curr.next = prev
                prev = curr
                curr = nxt
            return prev
        
        dummy = ListNode(0, head)
        prev = dummy
        while True:
            start = prev.next
            end = prev
            for _ in range(k):
                end = end.next
                if not end:
                    return dummy.next
            
            next_group = end.next
            new_head = reverse(start, end, next_group)
            prev.next = new_head
            start.next = next_group
            prev = start

    # NOTE: 分离两个链表
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':
        """ 138
        https://leetcode.cn/problems/copy-list-with-random-pointer/description/?envType=study-plan-v2&envId=top-100-liked
        """
        if not head:
            return None
        
        cur = head
        while cur:
            new_node = Node(cur.val)
            new_node.next = cur.next
            cur.next = new_node
            cur = new_node.next

        cur = head
        while cur:
            if cur.random:
                cur.next.random = cur.random.next
            cur = cur.next.next
        
        # 分离两个链表
        cur = head
        new_head = head.next
        copy_cur = new_head
        while cur:
            cur.next = copy_cur.next
            cur = cur.next
            if cur:
                copy_cur.next = cur.next
                copy_cur = copy_cur.next
        return new_head

    # NOTE: 多加练习，很难，链表的归并排序
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """ 148
        https://leetcode.cn/problems/sort-list/?envType=study-plan-v2&envId=top-100-liked
        """

        def split(self, head: ListNode, step: int)->ListNode:
            if not head:
                return None
            for _ in range(step-1):
                if head.next:
                    head = head.next
                else:
                    break
            second = head.next
            head.next = None
            return second
        
        def merge(self, l1:ListNode, l2:ListNode) -> ListNode:
            dummy = ListNode(0)
            cur = dummy
            while l1 and l2:
                if l1.val < l2.val:
                    cur.next = l1
                    l1 = l1.next
                else:
                    cur.next = l2
                    l2 = l2.next
                cur = cur.next
            cur.next = l1 if l1 else l2
            return dummy.next

        if not head or not head.next:
            return head

        length = 0
        node = head
        while node:
            length += 1
            node = node.next

        dummy = ListNode(0, head)
        step = 1
        while step < length:
            prev, curr = dummy, dummy.next
            while curr:
                left = curr
                right = split(left, step)
                curr = split(right, step)
                prev.next = self.merge(left, right)
                while prev.next:
                    prev = prev.next
            step <<= 1
        return dummy.next

    # NOTE: 多加练习，优先队列
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        """ 23
        https://leetcode.cn/problems/merge-k-sorted-lists/?envType=study-plan-v2&envId=top-100-liked
        """
        import heapq
        dummy = ListNode(0)
        cur = dummy
        heap = []
        for i, node in enumerate(lists):
            if node:
                heapq.heappush(heap, (node.val, i, node))
        while heap:
            val, i, node = heapq.heappop(heap)
            cur.next = node
            cur = cur.next
            if node.next:
                heapq.heappush(heap, (node.next.val, i, node.next))
        return dummy.next

""" 146 - LRU 最近最少使用 缓存 
https://leetcode.cn/problems/lru-cache/?envType=study-plan-v2&envId=top-100-liked
"""
   
class DLinkedNode:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {}
        # 使用伪头部和伪尾部节点
        self.head = DLinkedNode()
        self.tail = DLinkedNode()
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove_node(self, node: DLinkedNode):
        """移除节点"""
        node.prev.next = node.next
        node.next.prev = node.prev

    def _add_to_tail(self, node: DLinkedNode):
        """将节点添加到尾部（最近使用）"""
        node.prev = self.tail.prev
        node.next = self.tail
        self.tail.prev.next = node
        self.tail.prev = node

    def _move_to_tail(self, node: DLinkedNode):
        """将节点移动到尾部"""
        self._remove_node(node)
        self._add_to_tail(node)

    def get(self, key: int) -> int:
        if key in self.cache:
            node = self.cache[key]
            self._move_to_tail(node)
            return node.value
        return -1

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self._move_to_tail(node)
        else:
            if len(self.cache) >= self.capacity:
                # 删除最久未使用的节点（头部节点）
                removed = self.head.next
                self._remove_node(removed)
                del self.cache[removed.key]
            # 创建新节点并加入尾部
            new_node = DLinkedNode(key, value)
            self.cache[key] = new_node
            self._add_to_tail(new_node)

# Your LRUCache object will be instantiated and called as such:
# obj = LRUCache(capacity)
# param_1 = obj.get(key)
# obj.put(key,value)
```

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# ---------- 前序遍历 (根 -> 左 -> 右) ----------
def preorder_recursive(root, result=None):
    if result is None: result = []
    if root:
        result.append(root.val) # <- operation
        preorder_recursive(root.left, result)
        preorder_recursive(root.right,result)
    return result

def preorder_iterative(root):
    if not root: return []
    stack, result = [root], []
    while stack:
        node = stack.pop()
        result.append(node.val) # <- operation
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return result

# ---------- 中序遍历 (左 -> 根 -> 右) ----------
def inorder_recursive(root, result=None):
    if result is None: result = []
    if root:
        inorder_recursive(root.left, result)
        result.append(root.val) # <- operation
        inorder_recursive(root.right, result)
    return result

def inorder_iterative(root):
    stack, result = [], []
    curr = root
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        result.append(curr.val) # <- operation
        curr = curr.right
    return result

# ---------- 后序遍历 (左 -> 右 -> 根) ----------
def postorder_recursive(root, result=None):
    if result is None: result = []
    if root:
        postorder_recursive(root.left, result)
        postorder_recursive(root.right, result)
        result.append(root.val) # <- operation
    return result

def postorder_iterative_one_stack(root):
    stack, result = [], []
    curr = root
    last_visited = None
    while curr or stack:
        while curr:
            stack.append(curr)
            curr = curr.left
            
        peek = stack[-1]
        if peek.right and last_visited != peek.right:
            curr = peek.right
        else:
            result.append(peek.val) # <- operation
            last_visited = stack.pop()
    return result

# ---------- 层序遍历 (BFS) ----------
def level_order(root):
    if not root: return []
    
    from collections import deque
    result = []
    queue = deque([root])
    while queue:

        level_size = len(queue)
        level_vals = []
        for _ in range(level_size):
            node = queue.popleft()
            level_vals.append(node.val)
            if node.left: queue.append(node.left)
            if node.right: queue.append(node.right)
        
        result.append(level_vals) # <- operation
    return result

class Solution:
    """ Tree """

    # NOTE: 基础
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        """ 94 中序遍历: 左子树 -> 根节点 -> 右子树
        https://leetcode.cn/problems/binary-tree-inorder-traversal/description/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        stack = []
        cur = root
        while cur or stack:
            while cur:
                stack.append(cur)
                cur = cur.left
            cur = stack.pop()
            res.append(cur.val)
            cur = cur.right
        return res
    
    # NOTE: 树的解法: 递归 + 栈 + 队列 + DFS + BFS
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        """ 104
        https://leetcode.cn/problems/maximum-depth-of-binary-tree/description/?envType=study-plan-v2&envId=top-100-liked
        """

        # DFS
        if not root:
            return 0
        
        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)
        return 1 + max(left_depth, right_depth)

        stack = [(root, 1)]
        max_depth = 0
        while stack:
            node, depth = stack.pop()
            max_depth = max(max_depth, depth)
            if node.right:
                stack.append((node.right, depth+1))
            if node.left:
                stack.append((node.left, depth+1))
        return max_depth
    
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        """ 226
        https://leetcode.cn/problems/invert-binary-tree/?envType=study-plan-v2&envId=top-100-liked
        """
        if not root:
            return None
        
        root.left, root.right = root.right, root.left
        self.invertTree(root.left)
        self.invertTree(root.right)
        return root
    
        stack = [root]
        while stack:
            node = stack.pop()
            node.left, node.right = node.right, node.left
            if node.right:
                stack.append(node.right)
            if node.left:
                stack.append(node.left)
        return root
    
    def isSymmetric(self, root: Optional[TreeNode]) -> bool:
        """ 101
        https://leetcode.cn/problems/symmetric-tree/?envType=study-plan-v2&envId=top-100-liked
        """
        if not root:
            return True
        
        def check(left: TreeNode, right: TreeNode) -> bool:
            if not left and not right:
                return True
            if not left or not right:
                return False
            return left.val == right.val and check(left.left, right.right) and check(left.right, right.left)
            
        return check(root.left, root.right)
    
        stack = [(root.left, root.right)]
        while stack:
            left, right = stack.pop()
            if not left and not right:
                continue
            if not left or not right:
                return False
            if left.val != right.val:
                return False
            stack.append((left.left, right.right))
            stack.append((left.right, right.left))
        return True
        
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        """ 543
        https://leetcode.cn/problems/diameter-of-binary-tree/description/?envType=study-plan-v2&envId=top-100-liked
        """
        self.diameter = 0
        def height(node):
            if not node:
                return 0
            left_h = height(node.left)
            right_h = height(node.right)
            self.diameter = max(self.diameter, left_h + right_h)
            return 1 + max(left_h, right_h)
        height(root)
        return self.diameter
    
        diameter = 0
        height = {}
        stack = [(root, False)]
        while stack:
            node, visited = stack.pop()
            if not node:
                continue
            if not visited:
                stack.append((node, True))
                if node.right:
                    stack.append((node.right, False))
                if node.left:
                    stack.append((node.left, False))
            else:
                left_h = height.get(node.left, 0)
                right_h = height.get(node.right, 0)
                diameter = max(diameter, left_h + right_h)
                height[node] = 1 + max(left_h, right_h)
        return diameter
    
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        """ 102
        https://leetcode.cn/problems/binary-tree-level-order-traversal/?envType=study-plan-v2&envId=top-100-liked
        """
        if not root:
            return []
        
        from collections import deque
        res = []
        queue = deque([root])
        while queue:
            level_size = len(queue)
            level_vals = []
            for _ in range(level_size):
                node = queue.popleft()
                level_vals.append(node.val)
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
            res.append(level_vals)
        return res
    
    # NOTE: BST的中序遍历就是升序
    def sortedArrayToBST(self, nums: List[int]) -> Optional[TreeNode]:
        """ 108
        https://leetcode.cn/problems/convert-sorted-array-to-binary-search-tree/?envType=study-plan-v2&envId=top-100-liked
        """
        def build(left: int, right: int) -> Optional[TreeNode]:
            if left > right:
                return None
            mid = (left + right) // 2  # 选择中间元素作为根
            root = TreeNode(nums[mid])
            root.left = build(left, mid - 1)
            root.right = build(mid + 1, right)
            return root
        
        return build(0, len(nums) - 1)

        if not nums:
            return None

        mid = (0 + len(nums)-1)//2
        root = TreeNode(nums[mid])
        stack = []
        if mid+1 <= len(nums)-1:
            stack.append((mid+1, len(nums)-1, root, False))
        if 0 <= mid-1:
            stack.append((0, mid-1, root, True))

        while stack:
            left, right, parent, isLeft = stack.pop()
            mid = (left + right) // 2
            node = TreeNode(nums[mid])

            if isLeft:
                parent.left = node
            else:
                parent.right = node
            
            if mid + 1 <= right:
                stack.append((mid+1, right, node, False))
            if left <= mid-1:
                stack.append((left, mid-1, node, True))
        return root
    
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        """ 98
        https://leetcode.cn/problems/validate-binary-search-tree/description/?envType=study-plan-v2&envId=top-100-liked
        """
        self.prev = float('-inf')
        def inorder(node):
            if not node:
                return True
            if not inorder(node.left):
                return False
            if node.val <= self.prev:
                return False
            self.prev = node.val
            return inorder(node.right)
        return inorder(root)
    
        stack = []
        cur = root
        prev = float('-inf')
        while cur or stack:
            while cur:
                stack.append(cur)
                cur = cur.left
            cur = stack.pop()
            if cur.val <= prev:
                return False
            prev = cur.val
            cur = cur.right
        return True
    
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:
        """ 230
        https://leetcode.cn/problems/kth-smallest-element-in-a-bst/?envType=study-plan-v2&envId=top-100-liked
        """
        self.count = 0
        self.result = 0
        
        def inorder(node):
            if not node:
                return
            inorder(node.left)
            self.count += 1
            if self.count == k:
                self.result = node.val
                return
            inorder(node.right)

        inorder(root)
        return self.result

        stack = []
        cur = root
        count = 0
        while cur or stack:
            while cur:
                stack.append(cur)
                cur = cur.left
            cur = stack.pop()
            count += 1
            if count == k:
                return cur.val
            cur = cur.right
        return -1  # 理论上不会走到这里

    # 该层的第一个节点
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        """ 199 根节点 -> 右子树 -> 左子树
        https://leetcode.cn/problems/binary-tree-right-side-view/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        def dfs(node, depth):
            if not node:
                return
            if depth == len(res): # 巧妙: 该层的第一个节点
                res.append(node.val)
            dfs(node.right, depth+1)
            dfs(node.left, depth+1)
        dfs(root, 0)
        return res
        
        res = []
        stack = [(root, 0)]  # (节点, 深度)
        while stack:
            node, depth = stack.pop()
            # 如果当前深度等于结果长度，说明该层第一个访问的节点出现
            if depth == len(res):
                res.append(node.val)
            # 为了优先访问右子树，先压入左子节点，后压入右子节点
            if node.left:
                stack.append((node.left, depth + 1))
            if node.right:
                stack.append((node.right, depth + 1))
        return res
    
    # NOTE: 学习 Morris 遍历
    def flatten(self, root: Optional[TreeNode]) -> None:
        """ 114
        Do not return anything, modify root in-place instead.
        https://leetcode.cn/problems/binary-tree-right-side-view/description/?envType=study-plan-v2&envId=top-100-liked
        """

        if not root:
            return
        
        stack = [root]
        prev = None
        while stack:
            cur = stack.pop()
            if prev:
                prev.right = cur
                prev.left = None
            # 先压右，再压左，保证先处理左
            if cur.right:
                stack.append(cur.right)
            if cur.left:
                stack.append(cur.left)
            prev = cur

        # Morris: 利用了叶子节点的空闲右指针来保存回溯信息，避免了额外的栈空间
        cur = root
        while cur:
            if cur.left:
                # 找到左子树的最右节点
                predecessor = cur.left
                while predecessor.right:
                    predecessor = predecessor.right
                # 将原右子树接到前驱节点的右边
                predecessor.right = cur.right
                # 将左子树移到右边，左指针置空
                cur.right = cur.left
                cur.left = None
            # 继续处理下一个节点
            cur = cur.right
   
    # NOTE: 多加练习，再次练习
    def buildTree(self, preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
        """ 105
        https://leetcode.cn/problems/construct-binary-tree-from-preorder-and-inorder-traversal/?envType=study-plan-v2&envId=top-100-liked
        """
        if not preorder:
            return None
        
        # preorder: root, left, right
        # inorder: left, root, right
        root = TreeNode(preorder[0])
        stack = [root]
        in_index = 0
        for i in range(1, len(preorder)):
            node = stack[-1]

            if node.val != inorder[in_index]:
                node.left = TreeNode(preorder[i])
                stack.append(node.left)
            else:
                while stack and stack[-1].val == inorder[in_index]:
                    node = stack.pop()
                    in_index += 1
                node.right = TreeNode(preorder[i])
                stack.append(node.right)
        return root

    # NOTE: 树遍历 + 前缀和
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        """ 437
        https://leetcode.cn/problems/path-sum-iii/?envType=study-plan-v2&envId=top-100-liked
        """
        from collections import defaultdict
        prefix = defaultdict(int)
        prefix[0] = 1

        def dfs(node, curr_sum):
            if not node:
                return 0
            curr_sum += node.val

            count = prefix[curr_sum - targetSum]
            prefix[curr_sum] += 1
            count += dfs(node.left, curr_sum)
            count += dfs(node.right, curr_sum)
            prefix[curr_sum] -= 1
            return count
        return dfs(root, 0)
    
    # LCA 算法
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        """ 236
        https://leetcode.cn/problems/lowest-common-ancestor-of-a-binary-tree/submissions/708291640/?envType=study-plan-v2&envId=top-100-liked
        """
        if not root or root == p or root == q:
            return root
        left = self.lowestCommonAncestor(root.left, p, q)
        right = self.lowestCommonAncestor(root.right, p, q)
        if left and right:
            return root
        return left if left else right
    
        # 存储父节点
        parent = {root: None}
        stack = [root]
        while stack:
            node = stack.pop()
            if node.left:
                parent[node.left] = node
                stack.append(node.left)
            if node.right:
                parent[node.right] = node
                stack.append(node.right)
        ancestors = set()
        while p:
            ancestors.add(p)
            p = parent[p]
        while q not in ancestors:
            q = parent[q]
        return q

        # 两条路径
        def find_path(node, target, path):
            if not node:
                return False
            path.append(node)
            if node == target:
                return True
            if (node.left and find_path(node.left, target, path)) or (node.right and find_path(node.right, target, path)):
                return True
            path.pop()
            return False
        
        path_p, path_q = [], []
        find_path(root, p, path_p)
        find_path(root, q, path_q)
        lca = None
        for a, b in zip(path_p, path_q):
            if a == b:
                lca = a
            else:
                break
        return lca
    
    # NOTE: 多加练习，递归
    def maxPathSum(self, root: Optional[TreeNode]) -> int:
        """ 124
        https://leetcode.cn/problems/binary-tree-maximum-path-sum/?envType=study-plan-v2&envId=top-100-liked
        """
        self.max_sum = float('-inf')  # 全局最大路径和

        def max_gain(node):
            if not node:
                return 0
            # 递归计算左右子树的最大单侧路径和，负数则舍弃
            left_gain = max(max_gain(node.left), 0)
            right_gain = max(max_gain(node.right), 0)
            # 经过当前节点的路径和（可能拐弯）
            current_path_sum = node.val + left_gain + right_gain
            self.max_sum = max(self.max_sum, current_path_sum)
            # 返回当前节点的最大单侧路径和（用于父节点）
            return node.val + max(left_gain, right_gain)

        max_gain(root)
        return self.max_sum
```

```python
from collections import deque, defaultdict

"""
DFS: Stack[]
BFS: Queue
"""

class Solution:
    
    # NOTE: DFS、BFS
    def numIslands(self, grid: List[List[str]]) -> int:
        """ 200
        https://leetcode.cn/problems/number-of-islands/submissions/708426164/?envType=study-plan-v2&envId=top-100-liked
        """
        # DFS Stack
        if not grid:
            return 0
        m, n = len(grid), len(grid[0])
        count = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == '1':
                    stack = [(i, j)]
                    while stack:
                        x, y = stack.pop()
                        if x < 0 or x >= m or y < 0 or y >= n or grid[x][y] != '1':
                            continue
                        grid[x][y] = '0'
                        stack.append((x+1, y))
                        stack.append((x-1, y))
                        stack.append((x, y+1))
                        stack.append((x, y-1))
                    count += 1
        return count
    
        # DFS
        if not grid: return 0
        m, n = len(grid), len(grid[0])
        def dfs(i, j):
            if i < 0 or i >= m or j < 0 or j >= n or grid[i][j] != '1':
                return
            grid[i][j] = '0' # 标记已访问
            dfs(i+1, j)
            dfs(i-1, j)
            dfs(i, j+1)
            dfs(i, j-1)
        count = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == '1':
                    dfs(i, j)
                    count += 1
        return count
    
        # BFS deque
        from collections import deque
        if not grid:
            return 0
        m, n = len(grid), len(grid[0])
        count = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == '1':
                    queue = deque([(i, j)])
                    grid[i][j] = '0'
                    while queue:
                        x, y = queue.popleft()
                        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
                            nx, ny = x+dx, y+dy
                            if 0 <= nx < m and 0 <= ny < n and grid[nx][ny] == '1':
                                grid[nx][ny] = '0'
                                queue.append((nx, ny))
                    count += 1
        return count

    # NOTE: BFS
    def orangesRotting(self, grid: List[List[int]]) -> int:
        """ 994
        https://leetcode.cn/problems/rotting-oranges/?envType=study-plan-v2&envId=top-100-liked
        """
        row, col = len(grid), len(grid[0])
        fresh = 0
        queue = deque()

        for i in range(row):
            for j in range(col):
                if grid[i][j] == 2:
                    queue.append((i, j, 0))
                elif grid[i][j] == 1:
                    fresh += 1
        
        if fresh == 0:
            return 0
        
        time = 0
        directions = [(1,0), (-1,0), (0,1), (0,-1)]
        while queue:
            x, y, t = queue.popleft()
            time = max(time, t)
            for dx, dy in directions:
                nx, ny = x+dx, y+dy
                if 0 <= nx < row and 0 <= ny < col and grid[nx][ny] == 1:
                    grid[nx][ny] = 2
                    fresh -= 1
                    queue.append((nx, ny, t+1))
        return time if fresh == 0 else -1

    # NOTE: 拓扑排序 Kahn算法 或 DFS检测环
    def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:
        """ 207
        https://leetcode.cn/problems/course-schedule/?envType=study-plan-v2&envId=top-100-liked
        """
        # 构建邻接表
        graph = defaultdict(list)
        indegree = [0] * numCourses
        for a, b in prerequisites:
            graph[b].append(a)
            indegree[a] += 1
        
        # 队列存储入度为0的节点
        q = deque([i for i in range(numCourses) if indegree[i] == 0])
        count = 0
        while q:
            node = q.popleft()
            count += 1
            for nei in graph[node]:
                indegree[nei] -= 1
                if indegree[nei] == 0:
                    q.append(nei)
        return count == numCourses  

        # DFS
        graph = defaultdict(list)
        for a, b in prerequisites:
            graph[b].append(a)
        state = [0] * numCourses  # 0未访问，1访问中，2已访问

        def dfs(u):
            if state[u] == 1:
                return False  # 发现环
            if state[u] == 2:
                return True
            state[u] = 1
            for v in graph[u]:
                if not dfs(v):
                    return False
            state[u] = 2
            return True

        for i in range(numCourses):
            if not dfs(i):
                return False
        return True


    class Trie:
        def __init__(self):
            self.children = [None] * 26
            self.isEnd = False
        
        def searchPrefix(self, prefix: str) -> "Trie":
            node = self
            for ch in prefix:
                ch = ord(ch) - ord("a")
                if not node.children[ch]:
                    return None
                node = node.children[ch]
            return node

        def insert(self, word: str) -> None:
            node = self
            for ch in word:
                ch = ord(ch) - ord("a")
                if not node.children[ch]:
                    node.children[ch] = Trie()
                node = node.children[ch]
            node.isEnd = True

        def search(self, word: str) -> bool:
            node = self.searchPrefix(word)
            return node is not None and node.isEnd

        def startsWith(self, prefix: str) -> bool:
            return self.searchPrefix(prefix) is not None
```

```python
class Solution:

    def permute(self, nums: List[int]) -> List[List[int]]:
        """ 46
        https://leetcode.cn/problems/permutations/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        n = len(nums)
        used = [False] * n  # 标记数字是否已使用
        
        def backtrack(path: List[int]):
            if len(path) == n:
                res.append(path[:])  # 添加当前排列的副本
                return
            for i in range(n):
                if not used[i]:
                    used[i] = True
                    path.append(nums[i])
                    backtrack(path)
                    path.pop()      # 回溯
                    used[i] = False
        
        backtrack([])
        return res

        n = len(nums)
        res = []

        # (path, used_mask, start_idx)
        stack = [([], 0, 0)]
        while stack:
            path, used_mask, start = stack[-1]
            if len(path) == n:
                res.append(path)
                stack.pop()
                continue

            if start == n:
                stack.pop()
                continue

            i = start
            stack[-1] = (path, used_mask, i + 1)
            if not (used_mask >> i) & 1:
                new_path = path + [nums[i]]
                new_used = used_mask | (1 << i)
                stack.append((new_path, new_used, 0))
        
        return res
    

    def subsets(self, nums: List[int]) -> List[List[int]]:
        """ 78
        https://leetcode.cn/problems/subsets/submissions/710648979/?envType=study-plan-v2&envId=top-100-liked
        每个元素可以选择或不选择，递归生成所有子集。
        时间复杂度 O(n * 2^n)，空间复杂度 O(n)（递归栈深度）。
        """
        res = []
        n = len(nums)
        
        def backtrack(start: int, path: List[int]):
            res.append(path[:])
            for i in range(start, n):
                path.append(nums[i])
                backtrack(i + 1, path)
                path.pop()
        
        backtrack(0, [])
        return res
    
        """
        使用显式栈模拟回溯DFS, 不使用函数递归。
        每个栈帧保存当前索引和当前路径，当索引到达末尾时记录子集。
        """
        n = len(nums)
        res = []
        # 栈中元素: (index, path)
        stack = [(0, [])]

        while stack:
            idx, path = stack.pop()
            if idx == n:
                res.append(path)
                continue
            # 压入两个分支：不选当前元素 和 选当前元素
            # 由于栈是 LIFO，后压入的先处理，因此选的分支会先被处理（深度优先）
            stack.append((idx + 1, path[:]))          # 不选
            stack.append((idx + 1, path + [nums[idx]]))  # 选

        return res
    
    def letterCombinations(self, digits: str) -> List[str]:
        """ 17
        https://leetcode.cn/problems/letter-combinations-of-a-phone-number/submissions/710650952/?envType=study-plan-v2&envId=top-100-liked
        """  
        if not digits:
            return []
        
        mapping = {
            '2': 'abc', '3': 'def', '4': 'ghi', '5': 'jkl',
            '6': 'mno', '7': 'pqrs', '8': 'tuv', '9': 'wxyz'
        }
        res = []
        
        def backtrack(index: int, path: List[str]):
            if index == len(digits):
                res.append(''.join(path))
                return
            for ch in mapping[digits[index]]:
                path.append(ch)
                backtrack(index + 1, path)
                path.pop()
        
        backtrack(0, [])
        return res 
    
        if not digits:
            return []
        
        mapping = {
            '2': 'abc', '3': 'def', '4': 'ghi', '5': 'jkl',
            '6': 'mno', '7': 'pqrs', '8': 'tuv', '9': 'wxyz'
        }
        res = []
        # 栈元素: (当前组合字符串, 下一个要处理的数字索引)
        stack = [("", 0)]
        
        while stack:
            comb, idx = stack.pop()
            if idx == len(digits):
                res.append(comb)
                continue
            for ch in mapping[digits[idx]]:
                stack.append((comb + ch, idx + 1))
        
        return res

    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        """ 39
        https://leetcode.cn/problems/combination-sum/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        candidates.sort()  # 排序便于剪枝
        
        def backtrack(start: int, path: List[int], remain: int):
            if remain == 0:
                res.append(path[:])
                return
            for i in range(start, len(candidates)):
                if candidates[i] > remain:  # 剪枝
                    break
                path.append(candidates[i])
                backtrack(i, path, remain - candidates[i])  # 注意：允许重复使用，所以索引仍为 i
                path.pop()
        
        backtrack(0, [], target)
        return res
    
        candidates.sort()
        res = []
        # 栈元素: (start_index, path, remain)
        stack = [(0, [], target)]
        
        while stack:
            start, path, remain = stack.pop()
            if remain == 0:
                res.append(path[:])
                continue
            # 由于栈是 LIFO，为了保持顺序，我们从后向前遍历
            for i in range(start, len(candidates)):
                if candidates[i] > remain:
                    break
                # 将当前状态压栈（实际是压入选择当前数字后的状态）
                # 注意：因为可以重复选，下一个 start 仍然是 i
                stack.append((i, path + [candidates[i]], remain - candidates[i]))
        
        return res
    
    def generateParenthesis(self, n: int) -> List[str]:
        """ 22
        https://leetcode.cn/problems/generate-parentheses/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        
        def backtrack(path: str, left: int, right: int):
            if len(path) == 2 * n:
                res.append(path)
                return
            if left < n:
                backtrack(path + '(', left + 1, right)
            if right < left:
                backtrack(path + ')', left, right + 1)
        
        backtrack('', 0, 0)
        return res
    
        res = []
        # 栈元素: (当前字符串, 左括号使用数, 右括号使用数)
        stack = [('', 0, 0)]
        
        while stack:
            s, left, right = stack.pop()
            if len(s) == 2 * n:
                res.append(s)
                continue
            if left < n:
                stack.append((s + '(', left + 1, right))
            if right < left:
                stack.append((s + ')', left, right + 1))
        
        return res
    
    def partition(self, s: str) -> List[List[str]]:
        """ 131
        https://leetcode.cn/problems/palindrome-partitioning/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        n = len(s)
        
        def is_palindrome(left: int, right: int) -> bool:
            # 判断 s[left:right+1] 是否是回文
            while left < right:
                if s[left] != s[right]:
                    return False
                left += 1
                right -= 1
            return True
        
        def backtrack(start: int, path: List[str]):
            if start == n:
                res.append(path[:])
                return
            for end in range(start, n):
                if is_palindrome(start, end):
                    path.append(s[start:end+1])
                    backtrack(end + 1, path)
                    path.pop()
        
        backtrack(0, [])
        return res
    
        n = len(s)
        res = []
        
        def is_palindrome(left: int, right: int) -> bool:
            while left < right:
                if s[left] != s[right]:
                    return False
                left += 1
                right -= 1
            return True
        
        # 栈元素: (start_index, path)
        stack = [(0, [])]
        while stack:
            start, path = stack.pop()
            if start == n:
                res.append(path)
                continue
            # 为了保持字典序（或任意顺序），我们可以从后向前遍历，
            # 这样先压入的分支会后处理，但这里顺序不重要
            for end in range(start, n):
                if is_palindrome(start, end):
                    new_path = path + [s[start:end+1]]
                    stack.append((end + 1, new_path))
        
        return res

    """
    回溯 DFS N皇后
    """
        
    def exist(self, board: List[List[str]], word: str) -> bool:
        """ 79
        https://leetcode.cn/problems/word-search/?envType=study-plan-v2&envId=top-100-liked 
        """

        if not board or not board[0]:
            return False
        
        m, n = len(board), len(board[0])
        directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
        
        def dfs(r: int, c: int, idx: int) -> bool:
            if idx == len(word):
                return True
            if r < 0 or r >= m or c < 0 or c >= n or board[r][c] != word[idx]:
                return False
            
            # 标记当前格子已访问
            temp = board[r][c]
            board[r][c] = '#'
            
            for dr, dc in directions:
                if dfs(r + dr, c + dc, idx + 1):
                    return True
            
            # 恢复
            board[r][c] = temp
            return False
        
        for i in range(m):
            for j in range(n):
                if board[i][j] == word[0] and dfs(i, j, 0):
                    return True
        return False

        # 显式栈
        if not board or not board[0]:
            return False
        
        m, n = len(board), len(board[0])
        directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
        
        for i in range(m):
            for j in range(n):
                if board[i][j] != word[0]:
                    continue
                # 栈元素: (行, 列, 当前匹配索引, 访问状态)
                # 使用元组存储路径，每次扩展时新建状态
                stack = [(i, j, 0, [(i, j)])]  # (r, c, idx, visited_positions)
                while stack:
                    r, c, idx, visited = stack.pop()
                    if idx == len(word) - 1:
                        return True
                    for dr, dc in directions:
                        nr, nc = r + dr, c + dc
                        if 0 <= nr < m and 0 <= nc < n and (nr, nc) not in visited and board[nr][nc] == word[idx + 1]:
                            stack.append((nr, nc, idx + 1, visited + [(nr, nc)]))
        return False
    
    def solveNQueens(self, n: int) -> List[List[str]]:
        """
        https://leetcode.cn/problems/n-queens/?envType=study-plan-v2&envId=top-100-liked
        """
        res = []
        # 记录列、主对角线、副对角线的占用情况
        cols = [False] * n
        diag1 = [False] * (2 * n - 1)  # 主对角线：r - c 常数，范围 -(n-1) 到 (n-1)，偏移 n-1
        diag2 = [False] * (2 * n - 1)  # 副对角线：r + c 常数，范围 0 到 2n-2
        
        board = [['.'] * n for _ in range(n)]
        
        def backtrack(row: int):
            if row == n:
                # 将 board 转换为字符串列表
                res.append([''.join(row) for row in board])
                return
            for col in range(n):
                d1_idx = row - col + n - 1
                d2_idx = row + col
                if cols[col] or diag1[d1_idx] or diag2[d2_idx]:
                    continue
                # 放置皇后
                board[row][col] = 'Q'
                cols[col] = diag1[d1_idx] = diag2[d2_idx] = True
                backtrack(row + 1)
                # 撤销
                board[row][col] = '.'
                cols[col] = diag1[d1_idx] = diag2[d2_idx] = False
        
        backtrack(0)
        return res        

        res = []
        # 栈元素: (row, cols, diag1, diag2, board)
        # 因为状态较多，使用元组和列表的组合，但为了效率，我们只记录每行的皇后列位置
        # 使用位掩码优化（可选），这里用简单方式
        stack = [(0, [], [])]  # (row, col_positions, board_rows)
        # 其中 col_positions 为每行皇后所在的列号列表
        # board_rows 为已经构建好的行字符串列表
        
        while stack:
            row, cols, board_rows = stack.pop()
            if row == n:
                res.append(board_rows)
                continue
            for col in range(n):
                # 检查列冲突
                if col in cols:
                    continue
                # 检查对角线冲突
                conflict = False
                for r, c in enumerate(cols):
                    if abs(r - row) == abs(c - col):
                        conflict = True
                        break
                if conflict:
                    continue
                # 构建新行字符串
                new_row = ['.'] * n
                new_row[col] = 'Q'
                new_board_row = ''.join(new_row)
                stack.append((row + 1, cols + [col], board_rows + [new_board_row]))
        
        return res
```

## Part 5: Dynamic Programming

```python
class Solution:

    def climbStairs(self, n: int) -> int:
        """ 70
        https://leetcode.cn/problems/climbing-stairs/?envType=study-plan-v2&envId=top-100-liked
        """
        if n <= 1:
            return 1
        
        # dp[i] = dp[i-1] + dp[i-1] (i>=2)
        # dp[0], dp[1] = 1, 1
        prev, curr = 1, 1
        for i in range(2, n+1):
            next_step = prev + curr
            prev, curr = curr, next_step
        return curr
    
    def generate(self, numRows: int) -> List[List[int]]:
        """ 118
        https://leetcode.cn/problems/pascals-triangle/submissions/710012551/?envType=study-plan-v2&envId=top-100-liked
        """
        # dp[i][j] = dp[i-1][j-1] + dp[i-1][j]  
        # dp[0][0] = 1
        triangle = [[1]]
        for i in range(1, numRows):
            prev_row = triangle[i-1]
            curr_row = [1]
            for j in range(1, i):
                curr_row.append(prev_row[j-1] + prev_row[j])
            curr_row.append(1)
            triangle.append(curr_row)
        return triangle    
        
    def rob(self, nums: List[int]) -> int:
        """ 198
        https://leetcode.cn/problems/house-robber/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(nums)
        if n == 0:
            return 0
        if n == 1:
            return nums[0]

        # dp[i] = max(dp[i-1], dp[i-2] + nums[i]) (i >= 2)
        # dp[0], dp[1] = 0, nums[0]

        prev2, prev1 = nums[0], max(nums[0], nums[1])
        for i in range(2, n):
            curr = max(prev1, prev2 + nums[i])
            prev2 , prev1 = prev1, curr
        return prev1
    
    # NOTE: 完全背包问题
    def numSquares(self, n: int) -> int:
        """ 279
        https://leetcode.cn/problems/perfect-squares/?envType=study-plan-v2&envId=top-100-liked
        """
        # dp[i] = min(dp[i], dp[i-j*j]+1)  (for j with j*j <= i)
        # dp[0] = 0, dp[i] = inf or n+1
        dp = [n + 1] * (n + 1)
        dp[0] = 0
        squares = [i * i for i in range(1, int(n ** 0.5) + 1)]
        for i in range(1, n + 1):
            for square in squares:
                if square > i:
                    break
                dp[i] = min(dp[i], dp[i - square] + 1)
        return dp[n]
    
    def coinChange(self, coins: List[int], amount: int) -> int:
        """ 322
        https://leetcode.cn/problems/coin-change/?envType=study-plan-v2&envId=top-100-liked
        """
        # dp[i] = min(dp[i], dp[i - c] + 1) (c<=i)
        # return dp[amount]
        dp = [amount + 1] * (amount + 1)
        dp[0] = 0
        for i in range(1, amount + 1):
            for coin in coins:
                if coin <= i:
                    dp[i] = min(dp[i], dp[i - coin] + 1)
        return dp[amount] if dp[amount] != amount + 1 else -1

    # NOTE: 说破没酒喝
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        """ 139
        https://leetcode.cn/problems/word-break/?envType=study-plan-v2&envId=top-100-liked
        """
        word_set = set(wordDict)  # 转为集合，提高查找效率
        n = len(s)
        dp = [False] * (n + 1)
        dp[0] = True  # 空字符串可拆分

        for i in range(1, n + 1):
            for j in range(i):
                if dp[j] and s[j:i] in word_set:
                    dp[i] = True
                    break  # 一旦找到一种拆分即可停止
        
        return dp[n]
    
    def lengthOfLIS(self, nums: List[int]) -> int:
        """ 300
        https://leetcode.cn/problems/longest-increasing-subsequence/description/?envType=study-plan-v2&envId=top-100-liked
        """
        if not nums:
            return 0
        
        # dp[i] = max(dp[i], dp[j]+1) for nums[j]<nums[i]
        n = len(nums)
        dp = [1] * n
        for i in range(n):
            for j in range(i):
                if nums[j] < nums[i]:
                    dp[i] = max(dp[i], dp[j] + 1)
        return max(dp)
    
    # NOTE: 说破没酒喝，因为有负数，所以维护一个最小和最大
    def maxProduct(self, nums: List[int]) -> int:
        """ 152
        https://leetcode.cn/problems/maximum-product-subarray/?envType=study-plan-v2&envId=top-100-liked
        """

        # max_prod[i] = max(nums[i], max_prod[i-1] * nums[i], min_prod[i-1] * nums[i])
        # min_prod[i] = min(nums[i], max_prod[i-1] * nums[i], min_prod[i-1] * nums[i])

        if not nums:
            return 0
        max_prod = min_prod = result = nums[0]
        for i in range(1, len(nums)):
            # 保存之前的 max_prod 和 min_prod，因为当前计算会覆盖
            cur_max = max(nums[i], max_prod * nums[i], min_prod * nums[i])
            cur_min = min(nums[i], max_prod * nums[i], min_prod * nums[i])
            max_prod, min_prod = cur_max, cur_min
            result = max(result, max_prod)
        return result

    # NOTE: 0-1背包问题
    def canPartition(self, nums: List[int]) -> bool:
        """ 416
        https://leetcode.cn/problems/partition-equal-subset-sum/?envType=study-plan-v2&envId=top-100-liked
        将数组分成两个和相等的子集，等价于判断是否存在一个子集的和等于总和的一半
        """
        total = sum(nums)
        if total % 2 != 0:
            return False
        target = total // 2

        # dp[i] 代表 
        # dp[j] = dp[j] or dp[j - num]

        dp = [False] * (target+1)
        dp[0] = True
        for num in nums:
            for j in range(target, num-1, -1):
                dp[j] = dp[j] or dp[j-num]
        return dp[target]   


    """
    题目特征	          背包类型
    每个元素只能用一次	    0-1背包(如分割等和子集)
    每个元素可以用任意多次	完全背包（如零钱兑换、完全平方数）
    求能否恰好装满	       布尔型背包(or 运算）
    求最大价值 / 最小个数   最优型背包(max / min)

    状态转移方程的本质：
        0-1背包: dp[j] = max(dp[j], dp[j - w] + v)（内层倒序）
        完全背包: dp[j] = max(dp[j], dp[j - w] + v)（内层正序）
    记忆口诀: 0-1 背包倒着走，完全背包正着来。
    """

    def uniquePaths(self, m: int, n: int) -> int:
        """ 62
        https://leetcode.cn/problems/unique-paths/?envType=study-plan-v2&envId=top-100-liked
        """
        # dp[i][j] = dp[i-1][j] + dp[i][j-1]
        # dp[j] = dp[j] + dp[j-1]
        dp = [1] * n
        for i in range(1, m):
            for j in range(1, n):
                dp[j] += dp[j-1]
        return dp[-1]

    def minPathSum(self, grid: List[List[int]]) -> int:
        """ 64
        https://leetcode.cn/problems/minimum-path-sum/?envType=study-plan-v2&envId=top-100-liked
        """
        # dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1])
        # dp[0][j] = dp[0][j-1] + grid[0][j]
        # dp[i][0] = dp[i-1][0] + grid[i][0]
        m, n = len(grid), len(grid[0])
        # 第一行
        for j in range(1, n):
            grid[0][j] += grid[0][j-1]
        # 第一列
        for i in range(1, m):
            grid[i][0] += grid[i-1][0]
            # 其余位置
        for i in range(1, m):
            for j in range(1, n):
                grid[i][j] += min(grid[i-1][j], grid[i][j-1])
        return grid[m-1][n-1]
    
    # NOTE: 中心扩展 better
    def longestPalindrome(self, s: str) -> str:
        """ 5
        https://leetcode.cn/problems/longest-palindromic-substring/?envType=study-plan-v2&envId=top-100-liked
        """
        def expand(left, right):
            while left >= 0 and right < len(s) and s[left] == s[right]:
                left -= 1
                right += 1
            return s[left+1:right]
        
        res = ""
        for i in range(len(s)):
            # 奇数中心
            s1 = expand(i, i)
            # 偶数中心
            s2 = expand(i, i+1)
            res = max(res, s1, s2, key=len)
        return res
    
    # LCS 问题
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        """ 1143
        https://leetcode.cn/problems/longest-common-subsequence/?envType=study-plan-v2&envId=top-100-liked
        """    

        m, n = len(text1), len(text2)
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if text1[i-1] == text2[j-1]:
                    dp[i][j] = dp[i-1][j-1] + 1
                else:
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])
        return dp[m][n]
    
    def minDistance(self, word1: str, word2: str) -> int:
        """ 72
        https://leetcode.cn/problems/edit-distance/?envType=study-plan-v2&envId=top-100-liked
        """
        # 设 dp[i][j] 表示将 word1 的前 i 个字符转换成 word2 的前 j 个字符所需的最少操作数
        # dp[i][j] = dp[i-1][j-1]
        # dp[i][j] = min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1
        m, n = len(word1), len(word2)
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        for i in range(m + 1):
            dp[i][0] = i
        for j in range(n + 1):
            dp[0][j] = j

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if word1[i-1] == word2[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1
        return dp[m][n]
```

## Part 6: Greedy

```python
class Solution:

    """ 贪心算法 """
    def maxProfit(self, prices: List[int]) -> int:
        """ 121
        https://leetcode.cn/problems/best-time-to-buy-and-sell-stock/submissions/710524009/?envType=study-plan-v2&envId=top-100-liked
        """
        min_price = float('inf')
        max_profit = 0
        for price in prices:
            if price < min_price:
                min_price = price
            elif price - min_price > max_profit:
                max_profit = price - min_price
        
        return max_profit
    
    def canJump(self, nums: List[int]) -> bool:
        """ 55
        https://leetcode.cn/problems/jump-game/?envType=study-plan-v2&envId=top-100-liked
        """
        # max_reach = i+jump
        max_reach = 0
        for i, jump in enumerate(nums):
            if i > max_reach:
                return False
            max_reach = max(max_reach, i+jump)
            if max_reach >= len(nums)-1:
                return True
        return True
    
    # NOTE: 核心关键在判断下一跳能否到达，不能则jumps+=1    
    def jump(self, nums: List[int]) -> int:
        """ 45
        https://leetcode.cn/problems/jump-game-ii/?envType=study-plan-v2&envId=top-100-liked
        """
        n = len(nums)
        if n <= 1:
            return 0

        jumps = 0           # 跳跃次数
        current_end = 0     # 当前跳跃能到达的最远位置
        farthest = 0        # 下一步能到达的最远位置

        for i in range(n - 1):  # 不需要检查最后一个位置，因为到达即结束
            farthest = max(farthest, i + nums[i])
            # 当到达当前跳跃的边界时，必须进行下一次跳跃
            if i == current_end:
                jumps += 1   
                current_end = farthest
                # 如果已经可以到达末尾，提前结束
                if current_end >= n - 1:
                    break
        
        return jumps
    
    def partitionLabels(self, s: str) -> List[int]:
        """ 763
        https://leetcode.cn/problems/partition-labels/?envType=study-plan-v2&envId=top-100-liked
        贪心算法：记录每个字符最后出现的位置，遍历字符串，维护当前片段的结束位置。
        当遍历到结束位置时，当前片段结束，记录长度。
        """
        # 记录每个字符最后一次出现的索引
        last = {c: i for i, c in enumerate(s)}
        
        result = []
        start = 0
        end = 0
        
        for i, c in enumerate(s):
            # 更新当前片段的结束位置（取当前字符最后出现位置的最大值）
            end = max(end, last[c])
            # 如果当前位置等于结束位置，说明当前片段可以结束
            if i == end:
                result.append(end - start + 1)
                start = i + 1
        
        return result
```

## Part 7: Sorting and searching

```python
class Solution:

    """ 二分查找 """
    def searchInsert(self, nums: List[int], target: int) -> int:
        """ 35
        https://leetcode.cn/problems/search-insert-position/?envType=study-plan-v2&envId=top-100-liked
        二分查找，返回目标值应插入的位置。
        时间复杂度 O(log n)，空间复杂度 O(1)。
        """
        left, right = 0, len(nums) - 1
        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        return left
    
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        """ 74
        https://leetcode.cn/problems/search-a-2d-matrix/?envType=study-plan-v2&envId=top-100-liked
        将二维矩阵视为一个有序的一维数组进行二分查找。
        时间复杂度 O(log(m*n))，空间复杂度 O(1)。
        """
        if not matrix or not matrix[0]:
            return False
        
        m, n = len(matrix), len(matrix[0])
        left, right = 0, m * n - 1
        
        while left <= right:
            mid = (left + right) // 2
            # 将一维索引转换为二维坐标
            row = mid // n
            col = mid % n
            mid_val = matrix[row][col]
            
            if mid_val == target:
                return True
            elif mid_val < target:
                left = mid + 1
            else:
                right = mid - 1
        
        return False
    
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        """ 34
        https://leetcode.cn/problems/find-first-and-last-position-of-element-in-sorted-array/submissions/710605263/?envType=study-plan-v2&envId=top-100-liked
        使用两次二分查找分别找到第一个和最后一个位置。
        时间复杂度 O(log n)，空间复杂度 O(1)。
        """
        def findFirst(nums, target):
            left, right = 0, len(nums) - 1
            first = -1
            while left <= right:
                mid = (left + right) // 2
                if nums[mid] >= target:
                    right = mid - 1
                else:
                    left = mid + 1
                if nums[mid] == target:
                    first = mid
            return first
        
        def findLast(nums, target):
            left, right = 0, len(nums) - 1
            last = -1
            while left <= right:
                mid = (left + right) // 2
                if nums[mid] <= target:
                    left = mid + 1
                else:
                    right = mid - 1
                if nums[mid] == target:
                    last = mid
            return last
        
        return [findFirst(nums, target), findLast(nums, target)]        
    
    def search(self, nums: List[int], target: int) -> int:
        """ 33
        https://leetcode.cn/problems/search-in-rotated-sorted-array/submissions/710606686/?envType=study-plan-v2&envId=top-100-liked
        在旋转排序数组中二分查找目标值。
        时间复杂度 O(log n)，空间复杂度 O(1)。
        """
        left, right = 0, len(nums) - 1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                return mid
            
            # 判断左半部分是否有序
            if nums[left] <= nums[mid]:
                # 左半部分有序
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            else:
                # 右半部分有序
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        
        return -1
    
    def findMin(self, nums: List[int]) -> int:
        """ 153
        https://leetcode.cn/problems/find-minimum-in-rotated-sorted-array/?envType=study-plan-v2&envId=top-100-liked
        在旋转排序数组中找到最小值。
        使用二分查找，每次比较 mid 与 right 的值来判断最小值在哪一侧。
        时间复杂度 O(log n)，空间复杂度 O(1)。
        """
        left, right = 0, len(nums) - 1
        
        while left < right:
            mid = (left + right) // 2
            # 如果中间值大于右边值，说明最小值在右半部分
            if nums[mid] > nums[right]:
                left = mid + 1
            else:
                # 否则最小值在左半部分（包括 mid）
                right = mid
        
        return nums[left]     
    
    # NOTE: 二分查找必做题，多练习，注意边界情况的巧妙处理
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        """
        在 O(log(min(m, n))) 时间内找到两个有序数组的中位数。
        通过在较短的数组上进行二分查找，确定分割点。
        """
        # 确保 nums1 是较短的数组，以减少二分搜索的范围
        if len(nums1) > len(nums2):
            nums1, nums2 = nums2, nums1
        
        m, n = len(nums1), len(nums2)
        total_left = (m + n + 1) // 2  # 左半部分需要的元素个数
        
        left, right = 0, m
        while left <= right:
            i = (left + right) // 2  # nums1 的分割点，左边有 i 个元素
            j = total_left - i        # nums2 的分割点，左边有 j 个元素

            """
            在处理两个有序数组的分割问题时，边界情况指的是当分割点位于数组的起始或末尾时，一侧没有元素。
            此时我们不能直接取 `nums1[i-1]` 或 `nums1[i]`，否则会索引越界。
            因此，代码用 `float('-inf')` 和 `float('inf')` 来代表虚拟的边界值，保证比较逻辑的完整性。

                - **当 `i = 0`**：表示 `nums1` 中没有任何元素属于左半部分，那么左半部分的最大值应视为负无穷（`-inf`)
                    这样无论右半部分的 `nums2_left_max` 是什么，
                    条件 `nums1_left_max <= nums2_right_min` 都会成立（因为负无穷小于任何数），不会影响正确判断。
                - **当 `i = m`**：表示 `nums1` 中所有元素都属于左半部分，那么右半部分的最小值应视为正无穷（`inf`)
                    这样条件 `nums2_left_max <= nums1_right_min` 
                    总是成立（因为任何数都小于正无穷），同样不会干扰判断。

            对 `j` 也是同理，分别对应 `nums2` 的边界情况。

            在二分查找中，我们需要验证：
            nums1_left_max <= nums2_right_min 且 nums2_left_max <= nums1_right_min
            这是确定分割点正确的核心条件。
            当 `i=0` 时，`nums1_left_max` 不存在，但为了统一代码逻辑，我们给它赋一个极小的值，使第一个不等式自动满足；
            当 `i=m` 时，`nums1_right_min` 不存在，赋一个极大的值，使第二个不等式自动满足。
            这样，我们就不用单独写 `if` 判断来处理边界，代码简洁且逻辑一致。
            """

            # 处理边界情况，使用 -inf 和 inf 表示边界外的值
            nums1_left_max = nums1[i - 1] if i > 0 else float('-inf')
            nums1_right_min = nums1[i] if i < m else float('inf')
            nums2_left_max = nums2[j - 1] if j > 0 else float('-inf')
            nums2_right_min = nums2[j] if j < n else float('inf')
            
            if nums1_left_max <= nums2_right_min and nums2_left_max <= nums1_right_min:
                # 找到正确的分割点
                if (m + n) % 2 == 0:
                    return (max(nums1_left_max, nums2_left_max) + min(nums1_right_min, nums2_right_min)) / 2.0
                else:
                    return max(nums1_left_max, nums2_left_max)
            elif nums1_left_max > nums2_right_min:
                # i 太大，需要向左移动
                right = i - 1
            else:
                # i 太小，需要向右移动
                left = i + 1
        
        # 理论上不会执行到这里
        return -1.0

from typing import List

class SortingSolution:
    """排序算法集合（所有方法均返回新列表，不修改原列表）"""
    
    @staticmethod
    def quicksorting(nums: List[int]) -> List[int]:
        """快速排序（递归实现）"""
        if len(nums) <= 1:
            return nums[:]  # 返回副本
        pivot = nums[len(nums) // 2]
        left = [x for x in nums if x < pivot]
        middle = [x for x in nums if x == pivot]
        right = [x for x in nums if x > pivot]
        return SortingSolution.quicksorting(left) + middle + SortingSolution.quicksorting(right)

    @staticmethod
    def mergesorting(nums: List[int]) -> List[int]:
        """归并排序（分治法）"""
        if len(nums) <= 1:
            return nums[:]
        mid = len(nums) // 2
        left = SortingSolution.mergesorting(nums[:mid])
        right = SortingSolution.mergesorting(nums[mid:])
        return SortingSolution._merge(left, right)

    @staticmethod
    def _merge(left: List[int], right: List[int]) -> List[int]:
        """归并辅助函数"""
        result = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result

    @staticmethod
    def heapsorting(nums: List[int]) -> List[int]:
        """堆排序"""
        arr = nums[:]  # 复制原列表
        n = len(arr)
        # 建大顶堆
        for i in range(n // 2 - 1, -1, -1):
            SortingSolution._heapify(arr, n, i)
        # 逐个交换堆顶与末尾，并调整堆
        for i in range(n - 1, 0, -1):
            arr[i], arr[0] = arr[0], arr[i]
            SortingSolution._heapify(arr, i, 0)
        return arr

    @staticmethod
    def _heapify(arr: List[int], n: int, i: int) -> None:
        """堆调整辅助函数"""
        largest = i
        left = 2 * i + 1
        right = 2 * i + 2
        if left < n and arr[left] > arr[largest]:
            largest = left
        if right < n and arr[right] > arr[largest]:
            largest = right
        if largest != i:
            arr[i], arr[largest] = arr[largest], arr[i]
            SortingSolution._heapify(arr, n, largest)

    @staticmethod
    def bubblesorting(nums: List[int]) -> List[int]:
        """冒泡排序（优化版）"""
        arr = nums[:]
        n = len(arr)
        for i in range(n - 1):
            swapped = False
            for j in range(n - 1 - i):
                if arr[j] > arr[j + 1]:
                    arr[j], arr[j + 1] = arr[j + 1], arr[j]
                    swapped = True
            if not swapped:
                break
        return arr

    @staticmethod
    def insertionsorting(nums: List[int]) -> List[int]:
        """插入排序"""
        arr = nums[:]
        for i in range(1, len(arr)):
            key = arr[i]
            j = i - 1
            while j >= 0 and arr[j] > key:
                arr[j + 1] = arr[j]
                j -= 1
            arr[j + 1] = key
        return arr

    @staticmethod
    def selectionsorting(nums: List[int]) -> List[int]:
        """选择排序"""
        arr = nums[:]
        n = len(arr)
        for i in range(n - 1):
            min_idx = i
            for j in range(i + 1, n):
                if arr[j] < arr[min_idx]:
                    min_idx = j
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
        return arr
```



## 排序算法总结

### 1. 快速排序 (Quicksort)
- **思想**：分治法，选取基准元素（pivot），将数组分成小于、等于、大于三部分，递归排序。
- **时间复杂度**：平均 O(n log n)，最坏 O(n²)（当数组已有序且基准选择不当）。
- **空间复杂度**：O(log n)（递归栈），非原地实现 O(n)。
- **稳定性**：不稳定（交换可能改变相等元素的相对顺序）。

### 2. 归并排序 (Mergesort)
- **思想**：分治法，将数组分成两半，分别排序后合并。
- **时间复杂度**：始终 O(n log n)。
- **空间复杂度**：O(n)（需要额外数组存储合并结果）。
- **稳定性**：稳定（合并时保证相等元素原顺序）。

### 3. 堆排序 (Heapsort)
- **思想**：利用堆数据结构，先建堆（大顶堆），然后反复取出堆顶并调整。
- **时间复杂度**：始终 O(n log n)。
- **空间复杂度**：O(1)（原地排序）。
- **稳定性**：不稳定（堆调整可能打乱相等元素顺序）。

### 4. 冒泡排序 (Bubblesort)
- **思想**：相邻比较，每次将最大元素“冒泡”到末尾，优化版本可提前终止。
- **时间复杂度**：平均 O(n²)，最好 O(n)（已有序）。
- **空间复杂度**：O(1)。
- **稳定性**：稳定。

### 5. 插入排序 (Insertionsort)
- **思想**：将数组分为已排序和未排序部分，逐个将未排序元素插入正确位置。
- **时间复杂度**：平均 O(n²)，最好 O(n)（已有序）。
- **空间复杂度**：O(1)。
- **稳定性**：稳定。

### 6. 选择排序 (Selectionsort)
- **思想**：每次从未排序部分选出最小元素，放到已排序部分末尾。
- **时间复杂度**：始终 O(n²)。
- **空间复杂度**：O(1)。
- **稳定性**：不稳定（交换可能改变相等元素的顺序）。

| 算法       | 最好时间复杂度 | 平均时间复杂度 | 最坏时间复杂度 | 空间复杂度 | 稳定性 |
|------------|----------------|----------------|----------------|------------|--------|
| 快速排序   | O(n log n)     | O(n log n)     | O(n²)          | O(log n)   | 不稳定 |

**适用场景**：
- 数据规模小或基本有序：插入排序、冒泡排序。
- 数据规模大且对稳定性有要求：归并排序。
- 内存有限且不要求稳定：堆排序、快速排序（需优化基准选择避免最坏情况）。
- 一般情况下，快速排序综合性能最佳，但需注意最坏情况。

```python
class Solution:
    
    """ 前缀和 """

    # NOTE 数组遍历 + 前缀和
    def subarraySum(self, nums: List[int], k: int) -> int:
        """ 560
        https://leetcode.cn/problems/subarray-sum-equals-k/?envType=study-plan-v2&envId=top-100-liked
        """
        count = {0:1}
        pre = 0
        ans = 0
        for num in nums:
            pre += num
            ans += count.get(pre-k, 0)
            count[pre] = count.get(pre, 0)+1
        return ans
    
    # NOTE: 树遍历 + 前缀和
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        """ 437
        https://leetcode.cn/problems/path-sum-iii/?envType=study-plan-v2&envId=top-100-liked
        """
        from collections import defaultdict
        prefix = defaultdict(int)
        prefix[0] = 1

        def dfs(node, curr_sum):
            if not node:
                return 0
            curr_sum += node.val

            count = prefix[curr_sum - targetSum]
            prefix[curr_sum] += 1
            count += dfs(node.left, curr_sum)
            count += dfs(node.right, curr_sum)
            prefix[curr_sum] -= 1
            return count
        return dfs(root, 0)
    
    """ 异或 二进制"""

    def singleNumber(self, nums: List[int]) -> int:
        """ 136
        https://leetcode.cn/problems/single-number/?envType=study-plan-v2&envId=top-100-liked
        a ^ a = 0
        a ^ 0 = a
        """
        result = 0
        for num in nums:
            result ^= num
        return result
    
    def findDuplicate(self, nums: List[int]) -> int:
        """ 287
        https://leetcode.cn/problems/find-the-duplicate-number/submissions/710513408/?envType=study-plan-v2&envId=top-100-liked
        使用快慢指针(Floyd 判圈算法），将数组视为链表。
        由于数字范围 [1, n]，且长度为 n+1, 重复数字会导致环。
        时间复杂度 O(n)，空间复杂度 O(1)，且不修改原数组。
        """
        # 快慢指针初始都指向第一个元素
        slow = nums[0]
        fast = nums[0]
        
        # 第一阶段：找到环内的相遇点
        while True:
            slow = nums[slow]
            fast = nums[nums[fast]]
            if slow == fast:
                break
        
        # 第二阶段：找到环的入口（重复数）
        slow = nums[0]
        while slow != fast:
            slow = nums[slow]
            fast = nums[fast]
        
        return slow
        
    def majorityElement(self, nums: List[int]) -> int:
        """ 169
        使用摩尔投票算法Boyer-Moore Majority Vote Algorithm
        因为多数元素出现的次数 > n/2,所以可以两两抵消,最后剩下的就是多数元素。
        时间复杂度 O(n)，空间复杂度 O(1)
        """
        candidate = None
        count = 0
        
        for num in nums:
            if count == 0:
                candidate = num
            count += 1 if num == candidate else -1
        
        return candidate
    
    def sortColors(self, nums: List[int]) -> None:
        """ 75
        https://leetcode.cn/problems/sort-colors/description/?envType=study-plan-v2&envId=top-100-liked
        原地排序，使用荷兰国旗问题（三指针）算法。
        时间复杂度 O(n)，空间复杂度 O(1)。
        """
        # 三个指针：
        # left 指向下一个应该放 0 的位置
        # right 指向下一个应该放 2 的位置
        # curr 当前遍历的指针
        left, curr, right = 0, 0, len(nums) - 1
        
        while curr <= right:
            if nums[curr] == 0:
                # 将 0 交换到左边
                nums[left], nums[curr] = nums[curr], nums[left]
                left += 1
                curr += 1
            elif nums[curr] == 2:
                # 将 2 交换到右边
                nums[curr], nums[right] = nums[right], nums[curr]
                right -= 1
                # 注意：交换后 curr 不移动，因为新换来的值还需要检查
            else:  # nums[curr] == 1
                curr += 1


    # NOTE: 没看懂题目
    def nextPermutation(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        # 1. 从右向左找到第一个升序对 (i, i+1)，使得 nums[i] < nums[i+1]
        i = n - 2
        while i >= 0 and nums[i] >= nums[i+1]:
            i -= 1
        
        if i >= 0:
            # 2. 从右向左找到第一个大于 nums[i] 的数 nums[j]
            j = n - 1
            while j >= 0 and nums[j] <= nums[i]:
                j -= 1
            # 3. 交换 nums[i] 和 nums[j]
            nums[i], nums[j] = nums[j], nums[i]
        
        # 4. 反转 i+1 到末尾的部分，使其变为升序
        left, right = i + 1, n - 1
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
    
```
